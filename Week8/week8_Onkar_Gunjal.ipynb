# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
import logging
import math
import re

logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
logger = logging.getLogger("smart_agent")

#TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression safely."""
    try:
        cleaned_expression = expression.strip()
        if not cleaned_expression:
            return "Error in calculation"
        result = eval(cleaned_expression, {"__builtins__": {}}, {"math": math})
        return str(result)
    except Exception as exc:
        logger.error("Calculation failed: %s", exc)
        return "Error in calculation"
````

````python
#TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract meaningful keywords from text."""
    try:
        words = re.findall(r"\b[a-zA-Z]{3,}\b", text.lower())
        stop_words = {"the", "and", "from", "with", "that", "this", "into", "your", "what", "when", "where", "about", "have", "been"}
        keywords = [w for w in words if w not in stop_words]
        keywords = list(dict.fromkeys(keywords))
        return keywords[:5]
    except Exception as exc:
        logger.error("Keyword extraction failed: %s", exc)
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (IMPLEMENTED)

def agent(query: str):
    query_lower = query.lower().strip()

    if not query_lower:
        logger.warning("Empty query received")
        return {
            "type": "error",
            "result": "Query cannot be empty"
        }

    try:
        if "calculate" in query_lower or "compute" in query_lower or "evaluate" in query_lower:
            expression = re.sub(r"\b(calculate|compute|evaluate)\b", "", query_lower, count=1).strip()
            if not expression:
                logger.warning("Calculation request without expression")
                return {
                    "type": "error",
                    "result": "No expression provided for calculation"
                }
            logger.info("Routing to calculator for query: %s", query)
            return {
                "type": "calculation",
                "result": calculator(expression)
            }

        elif "keywords" in query_lower or "keyword" in query_lower or "extract" in query_lower:
            logger.info("Routing to keyword extractor for query: %s", query)
            return {
                "type": "keywords",
                "result": extract_keywords(query)
            }

        else:
            logger.info("Routing to general response for query: %s", query)
            return {
                "type": "general",
                "result": f"I can help with that. You asked: {query}"
            }

    except Exception as e:
        logger.exception("Unexpected error while processing query")
        return {
            "type": "error",
            "result": f"Error processing query: {str(e)}"
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```
